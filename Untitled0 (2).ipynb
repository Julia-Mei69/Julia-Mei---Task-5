```python
#1. Activate your toolkits
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load the built-in practice dataset
df = sns.load_dataset('tips')

# 3. Print a success message
print("Success! Your tools and data are ready.")
```

```python
# 1. See the raw data (the first 5 rows of spreadsheet)
print("--- FIRST 5 ROWS ---")
display(df.head())

# 2. Check the data structure (Info)
print("\n--- DATA INFO ---")
df.info()

# 3. Get statistical summaries (Describe)
print("\n--- STATISTICAL SUMMARY ---")
display(df.describe())
```

```python
 #1. Tell the software how big to make chart
plt.figure(figsize=(8, 5))

# 2. Create the distribution plot (histogram) for total bills
sns.histplot(data=df, x='total_bill', kde=True, color='skyblue')

# 3. Add titles so it looks like a professional report
plt.title('Distribution of Total Restaurant Bills', fontsize=14)
plt.xlabel('Total Bill Amount ($)', fontsize=12)
plt.ylabel('Number of Groups', fontsize=12)

# 4. Display the chart
plt.show()
```

```python
# 1. Arrange the data into a grid format (Days vs Smoker status)
grid_data = df.groupby(['smoker', 'day'], observed=False)['total_bill'].mean().unstack()

# 2. Tell the software how big to make chart
plt.figure(figsize=(8, 4))

# 3. Create the heatmap with colors and number labels
sns.heatmap(grid_data, annot=True, fmt=".2f", cmap="YlGnBu", cbar_kws={'label': 'Average Bill ($)'})

# 4. Add titles to make it look professional
plt.title('Average Total Bill by Day and Smoking Preference', fontsize=14, pad=15)
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Smoker Status', fontsize=12)

# 5. Display the chart
plt.show()
```

```python
#Bivariate Analysis (Comparing Total Bill across Days of the week)
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='day', y='total_bill', hue='sex', palette='Set2')
plt.title('Bivariate Analysis: Total Bill Distribution by Day and Gender', fontsize=14)
plt.xlabel('Day of the Week')
plt.ylabel('Total Bill ($)')
plt.show()
```

```python
#Multivariate Pairplot to inspect correlations and check for multicollinearity
sns.pairplot(df, hue='time', palette='coolwarm')
plt.suptitle('Multivariate Analysis: Pairplot Matrix by Meal Time', y=1.02)
plt.show()
```
